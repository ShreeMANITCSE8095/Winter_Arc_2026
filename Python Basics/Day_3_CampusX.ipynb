```python
import math
n= int(input("Enter a number: "))
ans= 0
for i in range(1, n+1):
    ans+= i/math.factorial(i)
print(ans)
```

```python
 # Negative Indexing in Python is used to access the elements of a sequence in reverse order. The last element of the sequence is accessed using the index -1, the second last element is accessed using the index -2, and so on.

s= "Hello World!"
print(s[-1])
```

```python
# Slicing in Python is used to access a range of elements in a sequence. It is done by specifying the start and end index of the sequence. The start index is inclusive and the end index is exclusive.

s= "Hello World!"
print(s[0:7]) # Print karega from 0th Index to (7-1)th Index tk
print(s[0:8:2]) # Print karega from 0th Index to (8-1)th Index tk with step size of 2
print(s[7:0:-1]) # Print karega from 7th Index to (0+1)th Index tk in reverse order
```

```python
s= "Hello"
print(s[::-1]) # Print karega from last Index to 0th Index tk in reverse order   
```
